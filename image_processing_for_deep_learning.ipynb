```python
# Importing the image module from matplotlib library

import matplotlib.image as mpimg
import matplotlib.pyplot as plt
```

```python
# loading the image through matplotlib.image

img = mpimg.imread("./practice_image/image_1.jpeg")
```

```python
type(img)
```

```python
# Checking the shape (dimension and color) of the image


img.shape
```

```python
print(img)
```

```python
# Converting a numpy array to an image

img_plot = plt.imshow(img)
plt.show()
```

```python
# resizing the image using pillow library

from PIL import  Image
```

```python
img = Image.open('./practice_image/image_1.jpeg')
img_resized = img.resize((200, 200))
```

```python
# Saving the resized image

img_resized.save("nedu_image_resized.jpg")
```

```python
img_plot = plt.imshow(img_resized)
plt.show()
```

# Converting an RGB image to a Grayscale Image using OpenCV





```python
# importing openCV library

%pip install opencv-python
import cv2
```

```python
img = cv2.imread("./nedu_image_resized.jpg")
```

```python
type(img)
```

```python
img.shape
```

```python
grayscale_img = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
```

```python
type(grayscale_img)
```

```python
grayscale_img.shape
```

```python
cv2.imshow("Grayscale image ", grayscale_img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# saving the grayscale image
cv2.imwrite('nedu_grayscale_image.jpg', grayscale_img)
```
